# Loan quote

Quotes a fixed-rate loan for an applicant.

```python
def loan_quote(amount: float, applicant: dict, term_months: int = 12,
               annual_rate: float = 0.06, tags: list[str] | None = None) -> dict:
    """Quote a fixed-rate loan and decide whether the applicant qualifies."""
    monthly_rate = annual_rate / 12
    payment = amount * monthly_rate / (1 - (1 + monthly_rate) ** -term_months)
    return {
        "monthly_payment": round(payment, 2),
        "approved": applicant["credit_score"] >= 650,
        "applicant_name": applicant["name"],
    }
```

```python
loan_quote(12000, {"name": "Ada", "credit_score": 720})
```
